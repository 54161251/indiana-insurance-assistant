```python
from IPython.display import display, Markdown
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain.output_parsers import StructuredOutputParser, ResponseSchema

llm = ChatOllama(
    model="huggingface.co/bartowski/Llama-3.2-1B-Instruct-GGUF",
    validate_model_on_init=True,
    num_predict=512,
)
```

```python
schema = [
    ResponseSchema(name="answer", description="answer to the math problem")
]
```

```python
parser = StructuredOutputParser.from_response_schemas(schema)
```

```python
prompt = PromptTemplate(
    template="""
    Slove the problem below \n\n
    Problem:{problem}\n\n
    {format_instructions}
    """,
    input_variables=["problem"],
    Partial_variables={"format_instructions": parser.get_format_instructions()}
)
```

```python
 parser.get_format_instructions()
```

```python
chain=prompt | llm | parser
```

```python
# chain.invoke({"problems": XXX + "/no_think"}) if Qwen 3
response = chain.invoke({"format_instructions": parser.get_format_instructions(), "problem": "what is 8 + 12 + 51"})
```

```python
type(response)
```

```python
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
```

```python
class MathAnswer(BaseModel):
    answer: float = Field(description = "answer to the math problem")
```

```python
parser = PydanticOutputParser(pydantic_object=MathAnswer)
```

```python
display(Markdown(parser.get_format_instructions()))
```

```python
chain=prompt | llm | parser
# chain.invoke({"problems": XXX + "/no_think"}) if Qwen 3
response = chain.invoke({"format_instructions": parser.get_format_instructions(), "problem": "what is 8 + 12 + 51"})
```

```python
schema2 = [
    ResponseSchema(name="answer_in_text", description="Translate text of the number")
]
parser2 = StructuredOutputParser.from_response_schemas(schema2)
```

```python
prompt2 = PromptTemplate(
    template="""
    Translate this numerical value to text. \n\n
    For example: 9 -> "Nine", 23 -> "Twenty three"
    Numerical value:{answer}\n\n
    {format_instructions}
    """,
    input_variables=["answer"],
    Partial_variables={"format_instructions": parser2.get_format_instructions()}
)
```

```python
prompt3 = PromptTemplate(
    template="""
    check if the answer is correct or not, or if it is the same with ground truth.\n\n
    Answer:{answer}\n\n
   Ground Truth:{ground truth}\n\n
    {format_instructions}
    """,
    input_variables=["answer", "ground_truth"],
    Partial_variables={"format_instructions": parser2.get_format_instructions()}
)
```

```python
chain = prompt1|llm|parser1
```

```python
response = chain.invoke({"problem": "what is 8 + 12 + 51"})
```
